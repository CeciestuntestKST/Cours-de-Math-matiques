```python
A=np.array([[1,3,2],[0,1,0],[-10,0,3]],dtype=int)
B=np.array([1,0,0,0,1,0,-2,1,1],dtype=int)
B=np.reshape(B,(3,3))
y=np.array([-1,2,1],dtype=int)
print(A+B)
print(A*B)
print(A@B)
print(A*y)
print(y*A)
print(A@y)
x=la.solve(A,y)
print(x)
la.norm(A@x-y)
```

```python
I3=np.diag([1,1,1],0)
D=I3-B@ np.transpose(B)

print(D)
```

```python
print(np.linalg.det(A))
print(np.linalg.det(D))
E=A@np.transpose(A)
print(np.linalg.det(E))
```

```python
Ai=la.solve(A,I3)
print (Ai)
Ei=la.solve(E,I3)
print(Ei)
```

```python
VaP=np.linalg.eigvals(E)
VeP=np.linalg.eig(E)
P=VeP[1]
print(P@np.transpose(P))
```

```python
Delta=np.diag(VeP[0])
print(E-P@Delta@np.transpose(P))
```

```python
def fa(F):
    return 2*F*F-3*F+1

def fb(F):
    return np.abs(fa(F))

def fc(F):
    return (F-1)*(F-2)

def fd(F):
    return 1/(1+F*F)

print(fa(A))
print(fb(A))
print(fc(A))
print(fd(A))
```

```python
def matrice(n):
    F=n*np.eye(n)
    for i in range (1,n):
        h=(n-i)*np.ones(n-i)
        F=F+np.diag(h,i)
        F=F+np.diag(h,-i)
    return F
matrice(10)
```
