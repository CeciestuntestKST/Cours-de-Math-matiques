```python
np.random.seed(1)
P=np.random.uniform(-1,1,(1000000,3))
fX=P[:,0]**2+P[:,1]**2+P[:,2]**2 <=1
print(fX)
Volume=np.mean(fX)*8
print(Volume)
```

```python
def Volume(n):
  P=np.random.uniform(-1,1,(1000000,n))
  somme=np.zeros(1000000)
  for i in range (n):
    somme+=P[:,i]**2
  fX=somme<=1
  Volume=np.mean(fX)*(2**n)
  return Volume
print(Volume(1))
print(Volume(2))
print(Volume(3))
print(Volume(4))
print(Volume(5))
```

6)a) Fonction de répartition.\
Pour $X$ suivant une loi de Bernouilli $\mathcal{B}(p)$ (proba $p$ que $X=1$ et proba $(1-p)$ que $X=0$).
$$F_X(t)=\begin{array}{ll} 0\text{ si $t<0$}\\ 1-p \text{ si $t\in ]0,1[$}\\ 1\text{ si $t\ge 1$}\end{array}  $$

6)b)La fonction de répartition d'une variable de loi exponentielle de paramètre 2 est croissante et continue sur $[0,+\infty]$:
$$F_X(t)=1-e^{-2x}$$
L'inverse généralisée est donc la réciproque de $F_X$:
$$F_X^{-1}(\omega)=-\frac{\ln(1-\omega)}{2}$$


```python
np.random.seed(1)
U=np.random.rand(10000)
def funinv(x):
  return -np.log(1-x)/2
X=funinv(U)
print(X)
```

7) Méthode de rejet

```python
def fun(x):
  return np.pi /2 * np.sin(np.pi*x)

c=np.pi/2

def rejet(n):
  Y=[]
  while (len(Y))<n:
    Y1 = np.random.rand()
    U = np.random.rand()
    if c*U>fun(Y1):
      Y.append(Y1)
  return Y

plt.hist(rejet(10000),bins=200,density=True,histtype='step')
x=np.linspace(0,1,10000)
plt.plot(x, np.pi/2*np.sin(np.pi*x),color='red')
```
